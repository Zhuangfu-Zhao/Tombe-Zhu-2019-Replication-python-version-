## 2. Eaton and Kortum (2002)

### Motivation

Classical trade theories (Ricardo, Heckscher-Ohlin), while extremely useful in highlighting the economic forces behind trade, are very difficult to generalize to a set-up with many trading partners and bilateral trade costs. Thus the classical theories do not provide much guidance in doing empirical work — they **lack quantifiability**.

> **The gravity equation** is one of the most successful empirical relationships in all of economics. Despite its empirical success, its lack of theoretical foundations and interpretability long limited its use for policy analysis or counterfactual predictions.

The EK model bridges this gap by introducing a **probabilistic representation of productivity differences** across countries and goods. It:

- Provides a **micro-founded derivation** of the gravity equation
- Naturally incorporates bilateral trade costs and multiple countries
- Yields elegant gravity expressions using a discrete-choice framework
- Shattered the belief that there couldn't be a Ricardian gravity model

### 2.1 Production

- Finite number of countries $i \in S = \{1, \dots, N\}$
- Continuum of goods $\Omega$
- Each country can produce every good but varies (exogenously) in productivity
- $z_i(\omega)$ — country $i$'s efficiency at producing good $\omega \in \Omega$

**Production function:**

$$y_i(\omega) = z_i(\omega) \, l_i(\omega) \tag{2.1}$$

> **Core Assumption:** $z_i(\omega)$ is drawn independently from a **Fréchet distribution** $F_i$ for any $\omega \in \Omega$:

$$F_i(z) = \exp(-T_i z^{-\theta})$$

where:

- **$T_i > 0$** — aggregate productivity (absolute advantage) of country $i$
- **$\theta > 1$** — dispersion of productivities (comparative advantage), constant across countries

> **Why Fréchet?**
>
> 1. **Theoretical:** If technology is determined by the best "idea" of how to produce, the limiting distribution is indeed Fréchet, where $T_i$ reflects the country's stock of ideas.
> 2. **Technical:** The logit model assumes Gumbel-distributed errors. If $X \sim \text{Fréchet}$, then $\ln X \sim \text{Gumbel}$. Thus: **Fréchet → log-linear models** (like gravity); **Gumbel → linear models**.

**Unit cost** of providing $\omega$ from country $i$ to $n$:

$$\text{Unit cost} = \frac{\tau_{ni}c_i}{z_i(\omega)}$$

where $c_i$ is the wage in country $i$, and $\tau_{ni}$ is the **iceberg trade cost** for shipping one unit from $i$ to $n$.

**Producer's optimization (perfect competition):**

$$\max_{q_{ni}(\omega)} \int_{\omega \in \Omega}\sum_{n \in S}\left(p_{ni}(\omega)-\frac{\tau_{ni}c_i}{z_i(\omega)}\right)q_{ni}(\omega) \, d\omega$$

subject to:
$$\sum_{n \in S} \tau_{ni}q_{ni}(\omega) \leq z_i(\omega) l_i(\omega), \quad \int_{\omega \in \Omega} l_i(\omega) \, d\omega \leq \bar{l}_i$$

Under perfect competition, for any $q_{ni}(\omega) > 0$:

$$p_{ni}(\omega) = \frac{\tau_{ni}c_i}{z_i(\omega)} \tag{2.2}$$

### 2.2 Preference

**Utility** of consumers in country $n$:

$$U_n = \left( \int_{\omega\in\Omega} q_n(\omega)^{\frac{\sigma-1}{\sigma}} \, d\omega \right)^{\frac{\sigma}{\sigma-1}} \tag{2.3}$$

**Dixit-Stiglitz price index:**

$$P_n = \left( \int_{\Omega} p_n(\omega)^{1-\sigma} \, d\omega \right)^{\frac{1}{1-\sigma}} \tag{2.4}$$

**Walrasian demand:**

$$q_n(\omega) = \frac{p_n(\omega)^{-\sigma}}{P_n^{1-\sigma}} \tag{2.5}$$

---

### 2.3 Four Key Results

#### 2.3.1 Price Distribution

Consumers in country $n$ purchase each good $\omega$ from the country offering the **lowest price**:

$$p_n(\omega) = \min_{i \in S} p_{ni}(\omega) = \min_{i \in S} \frac{\tau_{ni}c_i}{z_i(\omega)}$$

The probability that country $i$ offers country $n$ a price less than $p$:

$$G_{ni}(p) = 1 - \exp\left\{-p^\theta T_i(\tau_{ni}c_i)^{-\theta}\right\}$$

The probability that the **minimum** price in country $n$ is less than $p$:

$$G_n(p) = 1 - \exp\left\{-p^\theta\Phi_n\right\}$$

where:

$$\boxed{\Phi_n = \sum_{i \in S} T_i(\tau_{ni}c_i)^{-\theta}}$$

> **Key insight — $\Phi_n$:**
> $\Phi_n$ represents the **overall strength of global technologies** faced by country $n$. It condenses three dimensions of every potential supplier into a single scalar:
>
> - Productivity level $T_i$
> - Marginal cost $c_i$
> - Trade cost $\tau_{ni}$
>
> A larger $\Phi_n$ means country $n$ has access to more efficient supply → **lower prices**.
>
> Facing multiple suppliers yields the **same price distribution** as facing a single *"composite supplier"* with parameter $\Phi_n$.

> **Two extreme cases:**
>
> 1. **Free trade:** $\tau_{ni} = 1$ for all $n, i$ → $\Phi_n = \Phi$ is identical across all countries → same price distribution everywhere.
> 2. **Autarky:** $\tau_{ni} \to \infty$ for $n \neq i$ → $\Phi_n = T_n c_n^{-\theta}$.

---

#### 2.3.2 Price Index

Using the price index definition:

$$P_n = \left(\int_0^{\infty} p^{1-\sigma} \theta \Phi_n p^{\theta-1} \exp\left\{-p^\theta\Phi_n\right\} dp\right)^{\frac{1}{1-\sigma}}$$

Substituting $x \equiv p^\theta\Phi_n$:

$$\boxed{P_n = \gamma \, \Phi_n^{-\frac{1}{\theta}}}$$

where:

$$\gamma = \left(\Gamma\left(\frac{1-\sigma}{\theta} + 1\right)\right)^{\frac{1}{1-\sigma}}, \quad \theta > \sigma - 1$$

> **Result:** The price index is a simple power function of the technology parameter $\Phi_n$. Higher $\Phi_n$ → lower $P_n$.

---

#### 2.3.3 Allocation of Purchases

The probability that country $i$ is the **least-cost provider** to country $n$:

$$\boxed{\pi_{ni} = \frac{T_i (\tau_{ni} c_i)^{-\theta}}{\Phi_n}}$$

which is simply country $i$'s contribution to $\Phi_n$.

---

#### 2.3.4 Conditional Price Distribution

By the law of large numbers, $\pi_{ni}$ equals the fraction of goods that country $n$ purchases from country $i$.

> **Crucial property:** The distribution of prices of goods that country $n$ **actually purchases** from any country $i$ is the **same**:

$$\Pr\left\{ p_{ni}(\omega) \leq p \mid p_{ni}(\omega) \leq \min_{m \in S \setminus i} p_{nm}(\omega) \right\} = G_n(p)$$

> This greatly simplifies the expressions for trade flows — the price distribution conditional on sourcing from $i$ is the same as the unconditional distribution $G_n$.

---

**General Equilibrium:**

The equilibrium is composed of $\{q_n(\omega)\}_{n \in S, \omega \in \Omega}$, $\{p_n(\omega)\}_{n \in S, \omega \in \Omega}$, and $\{w_n\}_{n \in S}$ such that: 
\begin{align*}
    & p_n(\omega) = \min_{i \in S} \frac{\tau_{ni}w_i}{z_i(\omega)}, \\
    & q_n(\omega) = \frac{p_n(\omega)^{-\sigma}}{P_n^{1-\sigma}}X_n, \\
    & w_i = \max_{n \in S, \omega\in\Omega}\frac{p_n(\omega)z_i(\omega)}{\tau_{ni}}.
\end{align*}

Transform the system: 

\begin{align*}
    & p_n(\omega) q_n(\omega) = \frac{p_n(\omega)^{1-\sigma}}{P_n^{1-\sigma}}X_n, \\
    \rightarrow & \int_{\omega\in\Omega_{ni}} p_n(\omega) q_n(\omega) d\omega = \int_{\omega\in\Omega_{ni}} \frac{p_n(\omega)^{1-\sigma}}{P_n^{1-\sigma}} d\omega X_n, \\
    \rightarrow & X_{ni} = \pi_{ni} X_n. 
\end{align*}

\begin{align*}
    & w_i = \max_{n \in S, \omega\in\Omega}\frac{p_n(\omega)z_i(\omega)}{\tau_{ni}}, \\
    \rightarrow & w_iL_i = \sum_{n\in S} \int_{\omega\in\Omega_{ni}} \frac{p_n(\omega)z_i(\omega)l_i(\omega)}{\tau_{ni}} d\omega, \\
    \rightarrow & w_iL_i = \sum_{n\in S} \int_{\omega\in\Omega_{ni}} p_n(\omega)q_n(\omega) d\omega, \\
    \rightarrow & w_iL_i = \sum_{n\in S} \pi_{ni} X_n.
\end{align*}

With **trade balance**/budget constraint of household ($X_i = w_i L_i$):

$$\boxed{w_i L_i = \sum_{n \in S} \frac{T_i (\tau_{ni} w_i)^{-\theta}}{\sum_{m \in S} T_m(\tau_{nm}w_m)^{-\theta}} \, w_n L_n} \tag{2.7}$$

This provides a system of $N-1$ independent equations (Walras' Law) that determines wages $(w_1, \dots, w_N)$ up to a numeraire. Everything is as if countries were **exchanging labor** (Ricardian style).

Therefore, we can find $\{P_n\}_{n\in S}$, $\{X_n\}_{n\in S}$ and $\{w_n\}_{n\in S}$ such that: 
\begin{align*}
    & P_n = \gamma \, \left(\sum_{i \in S} T_i(\tau_{ni}w_i)^{-\theta}\right)^{-\frac{1}{\theta}}, \\
    & X_n = w_nL_n, \\
    & w_i L_i = \sum_{n \in S} \frac{T_i (\tau_{ni} w_i)^{-\theta}}{\sum_{m \in S} T_m(\tau_{nm}w_m)^{-\theta}} \, w_n L_n. 
\end{align*}

```python
import os
import numpy as np

SCRIPT_DIR = os.path.abspath('.')
RAW_DIR = os.path.join(SCRIPT_DIR, "data", "raw")
print("Path set successfully.")
```

```python
S = 5
theta = 4.0

np.random.seed(20260526)
T = np.random.uniform(1.0, 5.0, size=S)
L = np.random.uniform(10.0, 50.0, size=S)

tau = np.random.uniform(1.1, 2.0, size=(S, S))
np.fill_diagonal(tau, 1.0)

print(f"技术水平 (T):   {T}")
print(f"劳动力供给 (L): {L}\n")

w_1 = np.ones(S)

for it_num in range(1000):
    w_0 = w_1.copy()
    
    numer = T * (tau * w_0) ** (-theta)
    denom = numer.sum(axis=1, keepdims=True)
    pi = numer / denom
    
    X = w_0 * L
    
    w_1 = ((pi * X[:, None]).sum(axis=0) / L) * 0.2 + w_0 * 0.8
    w_1 = w_1 / w_1[0]

    if abs(w_1 - w_0).max() < 0.000001: 
        print(f"算法在第 {it_num} 次迭代时成功收敛。")
        break 

X_final = w_1 * L

print("\n--- 求解结果 ---")
print(f"均衡工资向量 (w): {w_1}")

# 验证劳动力市场是否真正出清 (销售额 vs 收入)
sales = (pi * X_final[:, None]).sum(axis=0)
print(f"每个国家的市场供需残差 (应接近 0): {sales - X_final}")
```
