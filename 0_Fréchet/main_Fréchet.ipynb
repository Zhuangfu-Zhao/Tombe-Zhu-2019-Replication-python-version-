## 1. General Gumbel \& General Fréchet

### 1.1 Gumbel Distribution
- Cumulative distribution function:
\begin{equation}
	\nonumber
	F(x) = \exp[-\exp(-\frac{x-\mu}{\sigma})].
\end{equation}
	- $\mu$ is the location parameter.
	- $\sigma > 0$ is the scale parameter.
	- $x\in R$.
- Summary Statistics:
	- Mean:
	\begin{equation}
		\nonumber
		E(x) = \mu + \sigma\gamma,
	\end{equation}
		- where $\gamma$ represents Euler's constant.
	- Variance:
	\begin{equation}
		\nonumber
		Var(x) = \frac{\pi^2}{6}\sigma^2.
	\end{equation}
- Standard Gumbel distribution: $\mu = 0$, $\sigma = 1$.
	- The cumulative distribution function becomes 
	\begin{equation}
		\nonumber
		F(x) = \exp[-\exp(-x)].
	\end{equation}
	- The mean becomes $\gamma$ and the variance $\pi^2/6$.

### 1.2 Fréchet Distribution
- Cumulative distribution function:
\begin{equation}
	\nonumber
	F(x) = \left\{
		\begin{aligned}
			&exp[-(\frac{x - \mu}{\sigma})^{-\alpha}], &\text{if}\ x > \mu; \\
			&0, &\text{if}\ x\le\mu.
		\end{aligned}
		\right.
\end{equation}
	- $\mu$ is the location parameter.
	- $\sigma > 0$ is the scale parameter.
	- $\alpha > 0$ is the shape parameter.
- Summary Statistics:
	- Mean: 
	\begin{equation}
		\nonumber
		E(x) = \mu + \sigma\Gamma(1 - \frac{1}{\alpha}),
	\end{equation}
		- where $\Gamma(a) \equiv \int_0^{+\infty}x^{a-1}e^{-x}dx$ is the Gamma function, $a > 0$.
		- Only if $\alpha > 1$ would the mean exist.
	- Variance:
	\begin{equation}
		\nonumber
		Var(x) = \sigma^2[\Gamma(1 - \frac{2}{\alpha}) - \Gamma^2(1 - \frac{1}{\alpha})].
	\end{equation}
		- Only if $\alpha > 2$ would the variance exist.
- Standard Fréchet distribution: $\mu = 0$, $\sigma = 1$.
	- The cumulative distribution function becomes 
	\begin{equation}
		\nonumber
		F(x) = \left\{
			\begin{aligned}
				&exp(-x^{-\alpha}), &\text{if}\ x > 0; \\
				&0, &\text{if}\ x\le 0.
			\end{aligned}
			\right.
	\end{equation}

## 2. Logit \& Eaton-Kortum Model

### 2.1 Logit Model (in brief)

- Utility that agent obtains from alternative $j$ is
\begin{equation}
    \nonumber
    U_{j} = V_{j} + \varepsilon_{j},
\end{equation}
    - where $\varepsilon_j$ is independently, identically distributed **standard Gumbel** across alternatives.
    - The cumulative distribution function of $\varepsilon_j$ is 
    \begin{equation}
        \nonumber
        F(\varepsilon) = \exp[-\exp(-\varepsilon)].
    \end{equation}
- The probability of choosing alternative $j$ is 
\begin{equation}
    \nonumber
    P_j = \frac{\exp(V_{in})}{\sum_j \exp(V_{jn})}.
\end{equation}

### 2.2 EK Model (in brief)

- The price of tradable good $\omega$ exporting from region $i$ to $n$ is 
\begin{equation}
    \nonumber
    p_{ni}(\omega) = \frac{c_i}{z_i(\omega)}\tau_{ni}.
\end{equation}
    - $z_i(\omega)$ is independently, identically distributed **Fréchet** across $\omega$ **within** region $i$.
    - However, $z_i(\omega)$ is not identically distributed **across** regions.
    - The cumulative distribution function of $z_i$ is 
    \begin{equation}
        \nonumber
        F_i(z) = \exp(-T_iz^{-\theta}),
    \end{equation}
        - where $T_i > 0$ measures the aggregate productivity of region $i$, and $\theta > 1$ governs the **inverse dispersion** (comparative advantage).
    - In this context, the Fréchet distribution is a special case with location parameter fixed at 0 and scale parameter set to $T_i^{1/\theta}$ ($\theta$ is the shape parameter): 
    \begin{equation}
        \nonumber
        F_i(z) = \exp(-T_iz^{-\theta}) = \exp[-(\frac{z}{T_i^{1/\theta}})^{-\theta}].
    \end{equation}

- The probability of $i$ being the lowest cost provider of good $\omega$ to $n$ is 
\begin{equation}
    \nonumber
    \pi_{ni} = \frac{T_i(c_i\tau_{ni})^{-\theta}}{\sum_k T_k(c_k\tau_{nk})^{-\theta}}.
\end{equation}
    - By LLN, this probability would be equal to the fraction of goods $i$ sells to $n$.

### 2.3 The Link between Gumbel \& Fréchet

If random variable $X$ is Fréchet distributed, $\ln X$ is distributed Gumbel.
- The statement is true only if the **location parameter is 0** in Fréchet distribution, which the Fréchet in EK satisfies.
- Loosely speaking, the Fréchet distribution works better for models that are log linear, whereas the Gumbel distribution works better for models that are linear.

Proof for the statement:
- Suppose that $X$ is Fréchet distributed (with $\mu = 0$),
\begin{equation}
	\nonumber
	F(x) = \exp\{-(\frac{x}{\sigma})^{-\alpha}\}.
\end{equation}
- Define $Y\equiv \ln X$. We have,
\begin{equation}
	\nonumber
	Pr\{Y\le y\} = Pr\{\ln X \le y\} = Pr\{X\le e^{y}\} = \exp\{-(\frac{e^y}{\sigma})^{-\alpha}\} = \exp\{-e^{-\alpha (y - \ln\sigma)}\}.
\end{equation}
- Define $\mu' \equiv \ln\sigma$ and $\sigma' \equiv 1/\alpha$, then we can see that $Y$ is distributed Gumbel:
\begin{equation}
	\nonumber
	Pr\{Y\le y\} = \exp[-\exp(-\frac{y - \mu'}{\sigma'})].
\end{equation}

## 3. Why Fréchet in EK?

- The production function in EK is in a log-linear form, where Fréchet distribution works better.
- The choice of Fréchet distribution is also justified. Kortum (1977) shows that, if the technology of producing goods is determined by the best idea of how to produce, then the limiting distribution is indeed Fréchet, where $T_i$ reflects the country's stock of ideas.
