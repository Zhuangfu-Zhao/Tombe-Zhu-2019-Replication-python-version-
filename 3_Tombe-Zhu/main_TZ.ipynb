## 1.  Model Setup

- $N + 1$ regions representing China’s $N$ provinces and the world.

- Each region has two sectors, $j \in \{ag, na\}$.

- Two continua of varieties $y^{ag}(\nu)$, $y^{na}(\nu)$.

- Each region-sector is endowed with a fixed factor (land) $\bar{S}_n^j$.

- There are $\bar{L}_n^j$ workers with hukou registration in region $n$ and sector $j$. 

---
### 1.1 Preferences
- Each worker has hukou registration in region $i$, sector $k$ and works in region $n$, sector $j$.

- Each maximizes a Cobb-Douglas utility: 
\begin{equation}
  \tag{1}
  u_n^j = \varepsilon_n^j \left[ \left(C_n^{j,\text{ag}}\right)^{\psi^{\text{ag}}} \left(C_n^{j,\text{na}}\right)^{\psi^{\text{na}}} \right]^\alpha \left(S_n^{j,h}\right)^{1-\alpha}, 
\end{equation}

  - subject to the budget constraint:
  \begin{equation*}
    P_n^{j,\text{ag}} C_n^{j,\text{ag}} + P_n^{j,\text{na}} C_n^{j,\text{na}} + r_n^j S_n^{j,h} \leq v_{in}^{kj},
  \end{equation*}

  - where:

    - $\psi^{\text{ag}} + \psi^{\text{na}} = 1$ : shares of final demand across the two sectors

    - $\alpha \in (0,1)$ : share of expenditure on goods ($1-\alpha$ on housing)

    - $\varepsilon_n^j$ : idiosyncratic preference shifter (i.i.d. across workers, sectors, regions)

- The **Final demand** can be expressed as:

\begin{equation}
  \tag{2}
  D_n^j = \alpha \psi^j \sum_{k \in \{\text{ag},\text{na}\}} v_n^k L_n^k
\end{equation}

\begin{equation*}
  r_n^jS_n^{j,h} = (1-\alpha) \sum_{k \in \{\text{ag},\text{na}\}} v_n^k L_n^k
\end{equation*}

### 1.2 Production, Trade, \& Prices

#### 1.2.1 Production Technology

- Agricultural and non-agricultural goods are CES composites of a continuum varieties:
\begin{equation*}
  Y_n^j = \left( \int_0^1 y_n^j(\nu)^{(\sigma-1)/\sigma} \, d\nu \right)^{\sigma/(\sigma-1)}.
\end{equation*}

  - **Input-output linkage** from CP model: $Y_n^j$ is directly consumed or used as intermediate inputs by producers of varieties

- The variety $y_n^j(\nu)$ is produced using labor, intermediate inputs, and land using Cobb-Douglas production technology

- For a firm with productivity $\varphi$ producing variety $\nu$ in region $n$, sector $j$, the marginal cost is:
\begin{equation}
  \tag{3}
  c_n^j(\varphi) \propto \frac{1}{\varphi} \left(w_n^j\right)^{\beta^j} \left(r_n^j\right)^{\eta^j} \prod_{k \in \{\text{ag},\text{na}\}} \left(P_n^k\right)^{\sigma^{jk}},
\end{equation}

  - where $\beta^j + \eta^j + \sum_{k \in \{\text{ag},\text{na}\}} \sigma^{jk} = 1$

    - $\beta^j$ : labor share in sector $j$ production

    - $\eta^j$ : land share in sector $j$ production

    - $\sigma^{jk}$ : intermediate input share from sector $k$ into sector $j$

> *To be noticed*: 
> - The marginal cost is **coded differently** from what is shown in the paper. The marginal cost is coded as 
\begin{equation*}
  c_n^j(\varphi) \propto \frac{1}{\varphi} \left(w_n^j\right)^{\beta^j} \left(r_n^j\right)^{\eta^j} [\prod_{k \in \{\text{ag},\text{na}\}} \left(P_n^k\right)^{\sigma^{jk}}]^{1 - \beta^j - \eta^j},
\end{equation*}
>   - so that $\sum_{k \in \{\text{ag},\text{na}\}} \sigma^{jk} = 1$
> - Typo: the model specification and the calibrated value for $\sigma$ (Table 3) are inconsistent

- Define $c_n^j \equiv \left(w_n^j\right)^{\beta^j} \left(r_n^j\right)^{\eta^j} \prod_{k \in \{\text{ag},\text{na}\}} \left(P_n^k\right)^{\sigma^{jk}}$, so that $c_n^j(\varphi) = c_n^j / \varphi$

- Assume that $\varphi$ is distributed **Fréchet** with CDF:
\begin{equation*}
  F_n^j(\varphi) = \exp\left(-T_n^j \varphi^{-\theta}\right),
\end{equation*}
  - where $T_n^j$ is the average productivity and $\theta$ governs dispersion (the trade elasticity).

#### 1.2.2 Trade Share \& Prices
- Shipping from region $i$ to $n$ incurs an **iceberg cost** $\tau_{ni}^j \geq 1$ (with $\tau_{nn}^j = 1$).

- Under perfect competition, we have
\begin{equation*}
  p_{ni}^j(\varphi) = \tau_{ni}^j \frac{c_i^j}{\varphi}
\end{equation*}

- Given that the producers of the composite goods purchase from the **cheapest** source, the equilibrium trade share and the corresponding price index are, respectively
\begin{equation}
  \tag{4}
  \boxed{
  \pi_{ni}^j = \frac{T_i^j \left(\tau_{ni}^j c_i^j\right)^{-\theta}}{\displaystyle \sum_{m=1}^{N+1} T_m^j \left(\tau_{nm}^j c_m^j\right)^{-\theta}}
  }
\end{equation}
  - where $\pi_{ni}^j$ is the share of region $n$'s spending on sector $j$ allocated to goods from region $i$.
\begin{equation}
  \tag{5}
  \boxed{
  P_n^j = \gamma \left[ \sum_{i=1}^{N+1} T_i^j \left(\tau_{ni}^j c_i^j\right)^{-\theta} \right]^{-1/\theta}
  }
\end{equation}
  - where $\gamma = \left[\Gamma\left(1 + \frac{1-\sigma}{\theta}\right)\right]^{\frac{1}{1-\sigma}}$ is a constant and $\Gamma(\cdot)$ is the Gamma function.

- **Total revenue** \& **total expenditure**:

  - Total revenue of sector $j$ producers in region $i$:
  \begin{equation}
    \tag{6}
    R_i^j = \sum_{n=1}^{N+1} \pi_{ni}^j X_n^j
  \end{equation}

  - Total spending on sector $j$ in region $n$ comprises final demand plus intermediate demand:
  \begin{equation}
    \tag{7}
    X_n^j = D_n^j + \sum_{k \in \{\text{ag},\text{na}\}} \sigma^{kj} R_n^k
  \end{equation}

> *To be noticed:*
> - Since the intermediate input share $\sigma$ is coded differently. The total expenditure is actually calculated by 
\begin{equation*}
    X_n^j = D_n^j + (1 - \beta^j - \eta^j)\sum_{k \in \{\text{ag},\text{na}\}} \sigma^{kj} R_n^k
\end{equation*}

### 1.3 Income of Workers

- Land is **collectively owned** by local hukou holders. 

- Only workers who remain in their hukou location receive fixed-factor income, while migrants receive only labor income.

- Land income

  - From Cobb-Douglas production shares, labor income, and the land market-clearing condition:
  \begin{align*}
    v_n^j L_n^j &= r_n^j \bar{S}_n^j + w_n^j L_n^j, \\
    \rightarrow r_n^j \bar{S}_n^j &= (1 - \alpha) v_n^j L_n^j + \eta^j R_n^j, \\
    \rightarrow v_n^j L_n^j &= (1 - \alpha) v_n^j L_n^j + \eta^j R_n^j + w_n^j L_n^j, \\
    \rightarrow v_n^j L_n^j &= (\eta^j R_n^j + \beta^j R_n^j) / \alpha.
  \end{align*}

  - Therefore, we can express both the total income $v_n^j L_n^j$ and the total income from land $r_n^j \bar{S}_n^j$ as functions of total labor income:
  \begin{equation*}
    v_n^j L_n^j = \frac{\eta^j + \beta^j}{\alpha \beta^j} \, w_n^j L_n^j
  \end{equation*}

  \begin{equation}
    \tag{8}
    r_n^j \bar{S}_n^j = [\frac{(1-\alpha)\beta^j + \eta^j}{\alpha \beta^j}] w_n^j L_n^j
  \end{equation}

- Income of workers 

  - The income of **local non-migrant** workers (registered and working in $(n,j)$):
  \begin{equation*}
    v_{nn}^{jj} = w_n^j + \frac{r_n^j \bar{S}_n^j}{\bar{L}_n^{jj}}
  \end{equation*}

  - The income of **migrant** workers (any worker with $n \neq i$ or $j \neq k$):
  \begin{equation*}
    v_{ni}^{jk} = w_i^k
  \end{equation*}

  - Effective rebate rate

    - Define $\delta_{ni}^{jk}$ as the ratio of actual income to labor income:
    \begin{equation}
      \tag{9}
      \delta_{ni}^{jk} = 
      \begin{cases}
        1 + \dfrac{(1-\alpha)\beta^j + \eta^j}{\alpha \beta^j} \cdot \dfrac{L_n^j}{\bar{L}_n^{jj}} & \text{if } n=i \text{ and } j=k \\
        1 & \text{if } n \neq i \text{ or } j \neq k
      \end{cases}
    \end{equation}

    - Then worker income is compactly written as $v_{ni}^{jk} = \delta_{ni}^{jk} \, w_i^k$.

### 1.4 Internal Migration

- Define the real wage index (which is calculated using prices indices and land rent)
\begin{equation*}
  V_i^k \equiv \frac{w_i^k}{\left[(P_i^{\text{ag}})^{\psi^{\text{ag}}} (P_i^{\text{na}})^{\psi^{\text{na}}}\right]^\alpha (r_i^k)^{1-\alpha}}
\end{equation*}

- Workers from $(n,j)$ choose destination $(i,k)$ to maximize their welfare:
\begin{equation*}
  \max_{i,k} \; \varepsilon_i^k \, \frac{\delta_{ni}^{jk} V_i^k}{\mu_{ni}^{jk}}
\end{equation*}
  - where $\mu_{ni}^{jk} \geq 1$ measures migration cost ($\mu_{nn}^{jj} = 1$)

- Assume that the idiosyncratic preference shifter is also distributed **Fréchet** with CDF:
\begin{equation*}
  F_\varepsilon(x) = \exp\left(-x^{-\kappa}\right)
\end{equation*}

  - where $\kappa > 1$ governs preference dispersion. Higher $\kappa \Rightarrow$ less dispersion, more elastic migration responses to income differences.

> *To be noticed:*
> - The Fréchet distribution used here is **standard** Fréchet, with zero location parameter and unit scale parameter.
> - It can be proved that this corresponds to Gumbel distribution with zero location parameter and scale parameter equaled to $1/\alpha$.

- Migration shares (proposition 1)

  - The share of workers registered in $(n,j)$ who choose to work in $(i,k)$ is:
  \begin{equation}
    \boxed{ 
    m_{ni}^{jk} = \frac{\left( \delta_{ni}^{jk} V_i^k / \mu_{ni}^{jk} \right)^\kappa}{\displaystyle \sum_{k' \in \{\text{ag},\text{na}\}} \sum_{i'=1}^{N+1} \left( \delta_{ni'}^{jk'} V_{i'}^{k'} / \mu_{ni'}^{jk'} \right)^\kappa} 
    } 
    \tag{10}
  \end{equation}

- The total employment at $(i,k)$ would be: 
\begin{equation*}
  L_i^k = \sum_{j \in \{\text{ag},\text{na}\}} \sum_{n=1}^{N} m_{ni}^{jk} \bar{L}_n^j
\end{equation*}
  - where $\bar{L}_n^j$ is the number of workers with hukou registration in $(n,j)$.

### 1.5 Solving the Model (Exact Hat-Algebra)

- Following Dekle, Eaton & Kortum (2007), the model solve for relative changes $\hat{x} \equiv x'/x$ from an initial to a counterfactual equilibrium

- Aggregate welfare and real GDP changes (proposition 2)

  - The change in aggregate welfare is 
  \begin{equation}
    \boxed{ 
    \hat{W} = \sum_{j \in \{\text{ag},\text{na}\}} \sum_{n=1}^{N} \omega_n^j \; \hat{V}_n^j \; \hat{\delta}_{nn}^{jj} \; \left( \hat{m}_{nn}^{jj} \right)^{-1/\kappa}
    } 
    \tag{11}
  \end{equation}
    - where $\omega_n^j \propto L_n^j V_n^j \delta_{nn}^{jj} (m_{nn}^{jj})^{-1/\kappa}$ is the initial contribution of region-sector $(n,j)$ to aggregate welfare.

  - The change in real GDP is 
  \begin{equation}
    \boxed{ 
    \hat{Y} = \sum_{j \in \{\text{ag},\text{na}\}} \sum_{n=1}^{N} \phi_n^j \; \hat{V}_n^j \; \hat{L}_n^j
    } 
    \tag{12}
  \end{equation}
    - where $\phi_n^j \propto V_n^j L_n^j$ is the initial contribution of $(n,j)$ to aggregate real GDP.
